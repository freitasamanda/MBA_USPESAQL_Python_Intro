# Criando funções próprias
Uma função é uma forma de simplificar um código e não ter que executá-lo várias vezes.

```python
def converter_milha(milha): #função que converte milha para km, o input é milha
    km = (milha * 1.6093)
    return f"O valor de {milha} milhas em quilômetros é {km}km"
```

```python
converter_milha(200)
```

```python
milhas_convertidas = pd.Series([60, 100, 20, 30])
```

```python
#aqui para cada valor eu vou usar um for para percorrer a Series inteira e ir printando um por um
for m in milhas_convertidas:
    print(converter_milha(m))
```

```python
# agora criando uma função com dois argumentos
def calcular_area(b, h):
    area = (b * h)
    return f"A área total é {area}m²"
```

```python
calcular_area(10, 12)
```

## Criando condições com if, elif e else

- if = se
- elif = se nao corresponder ao if, podemos fazer vários critérios
- else = se não, o restante, se todas as condições deram errado
- while = o for percorre o elemento base que você está estabelecendo enquanto o while tem uma condição de parada

```python
valor = 100
# condição if
if valor == 10**2:
    print("Valor Correto")
else:
    print("Valor Incorreto")

valor = 300

# condição if
if valor == 10**2:
    print("Valor Correto")
else:
    print("Valor Incorreto")
```

```python
# Usando com vários critérios
def verificar_salario(salario):
    if salario <=1621:
        print("O salário é menor que 1 salário mínimo")
    elif salario > 1621 and salario <= 4863:
        print("O salário está entre 1 e 3 salários mínimos")
    elif salario > 4863 and salario <= 8105:
        print("O salário está entre 3 e 5 salários mínimos")
    else:
        print("O salário é maior que 5 salários mínimos")

verificar_salario(4750)
```

```python
# Coeficiente de varição criando uma função

def coef_var(x):
    coeficiente = (np.std(x) / np.mean(x)) * 100
    coeficiente = np.round(coeficiente, decimals=2)
    return f"O coeficiente de variação é {coeficiente}%."
```

apliacando while em uma taxa de juros de 10% por ano até que o montante seja menor do que 10 mil

```python
saldo_investimento = 100
lista_invest = []

while saldo_investimento < 10000:
    saldo_investimento = (saldo_investimento*1.10)
    lista_invest.append(saldo_investimento)

print(lista_invest)
```
